# Digital High-Pass Filter

In this section, we are going to do the following activities:

- discretize a continuous-time high-pass filter by using bilinear transform (trapezoidal or Tustin) method
- implement the discretized high-pass filter into MATLAB Simulink
- compare the results from the implemented discrete high-pass filter to the shipped discrete high-pass filter in MATLAB Simulink

## Required Imports

````python
from IPython.core.display import HTML
from sympy import *
from mathprint import *
````

````python
Ts, tau = symbols('T_s tau', positive=True)
s = symbols('s', complex=True)
z = symbols('z')
omega    = symbols('omega'   , positive=True)
wc = symbols('omega_c', positive=True)

x, y = symbols('x y')
x0, x1, x2, x3 = symbols('x_k x_{k-1} x_{k-2} x_{k-3}')
y0, y1, y2, y3 = symbols('y_k y_{k-1} y_{k-2} y_{k-3}')
````

---


## First-Order High-Pass Filter
### The Complementary Form

Transfer function of a first-order highpass filter as a complementary to  a first-order lowpass filter(time-constant filter):

$$ H_{HP}(s) = 1 - H_{LP}$$


$$ H(s) = 1 - \frac{1}  {\tau s+1} $$

where $\tau$ is the filter time constant (in seconds).

__Discretization with Bilinear Transformation__

Next, we transform $s$ into $z$ by applying the following substitution.

$$ \frac{1}{s} \longleftarrow \frac{T_s}{2} \frac{z+1}{z-1} $$

````python
H = 1 - 1 / (tau*s+1)
mprint('H=',latex(H))

H = H.subs(1/s, Ts/2 * (z+1)/(z-1))
mprint('H=',latex(H))
````

Let us define $x$ as the input to the filter and $y$ as the output (filtered input).

````python
eq = Eq(y, H * x)
mprint(latex(eq))

eq = simplify(eq)
mprint(latex(eq))

eq = Eq(numer(eq.rhs), expand(eq.lhs * denom(eq.rhs)))
mprint(latex(eq))

eq =expand(Eq(numer(eq.rhs)/z/Ts, eq.lhs * denom(eq.rhs)/z/Ts))
mprint(latex(eq))
````

Apply the following substitutions:

* $y$ becomes $y_{k}$
* $y/z$ becomes $y_{k-1}$
* $x$ becomes $x_{k}$
* $x/z$ becomes $x_{k-1}$

````python
eq = eq.subs(x/z**3, x3).subs(x/z**2, x2).subs(x/z, x1).subs(x, x0).subs(y/z**3, y3).subs(y/z**2, y2).subs(y/z, y1).subs(y, y0)
mprint("\\small ", latex(eq))
````

Finally, by grouping the variables, we obtain:

````python
eq = Eq(collect(eq.rhs, [2,tau/Ts, x0, x1, x2, x3]), collect(eq.lhs, [y0, y1, y2, y3]) )
mprintb("\\small ", latex(eq))
````

## Second Order High-Pass Butterworth Filter

### The Complementary Form
Transfer function of a second order highpass filter as a complementary to a second-order lowpass filter (Butterworth):

$$ H_{HP}(s) = 1 - H_{LP}$$

$$ H(s) = 1 - \frac{\omega_{c}^{2}}{\omega_{c}^{2} + \sqrt{2} \omega_{c} s + s^{2}} $$

where $\omega_c$ is the analog cutoff angular frequency (in rad/s).


The denominator below is obtained by scaling the normalized Butterworth prototype $B_2(p)=p^2+\sqrt{2}p+1$. See {doc}`Butterworth Coefficient Table <butterworth-coefficients>` for this derivation.

**Important distinction:** The complementary filter developed here, $H_{HP}(s)=1-H_{LP}(s)$, is **not** the standard second-order Butterworth high-pass filter. The standard Butterworth high-pass response is

$$H_{HP,\mathrm{Butterworth}}(s)=\frac{s^2}{s^2+\sqrt{2}\omega_c s+\omega_c^2}.$$

Both designs reject DC, but their frequency responses differ. We retain the complementary derivation below as a separate useful construction.

__Discretization with Bilinear Transformation__

Similiar to the previous section, here we also transform $s$ into $z$ by applying the following substitution.

$$ \frac{1}{s} \longleftarrow \frac{T_s}{2} \frac{z+1}{z-1} $$

````python
H = 1 - wc**2 / (wc**2+sqrt(2)*s*wc+s**2)
mprint('H=',latex(H))

H = simplify(1 - wc**2 / (wc**2+sqrt(2)*s*wc+s**2))
mprint('H=',latex(H))

H = simplify(H.subs(1/s, Ts/2 * (z+1)/(z-1)))
mprint('H=',latex(H))
````

Let us define $x$ as the input to the filter and $y$ as the output (filtered input).

````python
eq = Eq(y, H * x)
mprint(latex(eq))

eq = simplify(eq)
mprint(latex(eq))

eq = Eq(numer(eq.rhs), expand(eq.lhs * denom(eq.rhs)))
mprint(latex(eq))

eq =expand(Eq(numer(eq.rhs)/z**2/Ts**2/wc**2, eq.lhs * denom(eq.rhs)/z**2/Ts**2/wc**2))
mprint(latex(eq))
````

Apply the following substitutions:

* $y$ becomes $y_{k}$
* $y/z$ becomes $y_{k-1}$
* $y/z^2$ becomes $y_{k-2}$
* $x$ becomes $x_{k}$
* $x/z$ becomes $x_{k-1}$
* $x/z^2$ becomes $x_{k-2}$

````python
eq = eq.subs(x/z**3, x3).subs(x/z**2, x2).subs(x/z, x1).subs(x, x0).subs(y/z**3, y3).subs(y/z**2, y2).subs(y/z, y1).subs(y, y0)
mprint(latex(eq))
````

Finally, by grouping the variables, we obtain:

````python
eq = Eq(collect(eq.rhs, [x0, x1, x2, x3]), collect(eq.lhs, [y0, y1, y2, y3]) )
mprintb(latex(eq))
````

__However, the complementary form does NOT have +40 dB/decade at $\omega=\omega_c$.__ We will show this through these steps:

- Substitute $j\omega$ to $s$ in $H(s)$ and convert it to dB. We name is as $M(j\omega)$.
- Calculate $S=\frac{dM}{d\omega}$

These steps will give us the slope in `db/(rad/s)`.

````python
H = simplify(1 - wc**2 / (wc**2+sqrt(2)*s*wc+s**2))
Hjw = H.subs(s, I*omega)

M = 20*log(abs(Hjw))/log(10)
mprint('M(j\\omega)=', latex(M))

dM = simplify(diff(M, omega))
mprint('S(\\omega)=\\frac{dM(\\omega)}{d\\omega}=', latex(dM))
````

However, the unit here is `dB/(rad/s)`, not `dB/decade`.
To obtain `dB/decade`:

````python
Sdb = simplify(omega * log(10) * dM)
mprint("S_{dB}(\\omega)=", latex(Sdb))
````

Let us take $\omega_c=1$ and plot the slope for arbitrary $\omega$.

````python
Sdb_wc1 = simplify(Sdb.subs(wc, 1))

mprint(
    r'S_{\mathrm{dB}}(\omega)\big|_{\omega_c=1}=',
    latex(Sdb_wc1)
)

p = plot(
    Sdb_wc1,
    (omega, 0.1, 100),
    size=(5, 2),
    show=False,
    title=r'$S_{\mathrm{dB}}(\omega),\quad \omega_c=1$',
    xlabel=r'$\omega$',
    ylabel=r'dB/decade',
    xscale='log'
)

p.show()
````

````python
Sdb_wc = simplify(Sdb.subs(omega, wc))
Sdb_wc
````

Thus, it is only about __6.7 dB / decade__! We can assure this by checking maximum an minimum possible slopes, which is 0 dB / decade and 20 db / decade, respectively.

````python
Sdb_low = simplify(limit(Sdb, omega, 0, dir='+'))
Sdb_high = simplify(limit(Sdb, omega, oo))

mprint(
    r'\lim_{\omega\to0}S_{\mathrm{dB}}(\omega)=',
    latex(Sdb_low)
)

mprint(
    r'\lim_{\omega\to\infty}S_{\mathrm{dB}}(\omega)=',
    latex(Sdb_high)
)
````

The complementary form has an asymptotic low-frequency slope of only +20 dB/decade, rather than the +40 dB/decade of a conventional second-order HPF. At the cutoff, its local slope is approximately +6.67 dB/decade.

### Standard Form

Since the complementary form only gives us 6.7dB / decade, while for a second order filter our target is 40 dB / decade, we must modify the filter equation slightly.

Infact, this is the standard form of a secon-dorder Butterworth highpass filter.

````python
H = s**2 / (wc**2+sqrt(2)*s*wc+s**2)
mprintb('H=',latex(H))
````

````python
H = simplify(H.subs(1/s, Ts/2 * (z+1)/(z-1)))
mprint('H=',latex(H))
````

````python
eq = Eq(y, H * x)
mprint(latex(eq))

eq = simplify(eq)
mprint(latex(eq))

eq = Eq(numer(eq.rhs), expand(eq.lhs * denom(eq.rhs)))
mprint(latex(eq))

eq =expand(Eq(numer(eq.rhs)/z**2/Ts**2/wc**2, eq.lhs * denom(eq.rhs)/z**2/Ts**2/wc**2))
mprint(latex(eq))
````

````python
eq = eq.subs(x/z**3, x3).subs(x/z**2, x2).subs(x/z, x1).subs(x, x0).subs(y/z**3, y3).subs(y/z**2, y2).subs(y/z, y1).subs(y, y0)
mprint(latex(eq))
````

````python
eq = Eq(collect(eq.rhs, [x0, x1, x2, x3]), collect(eq.lhs, [y0, y1, y2, y3]) )
mprintb(latex(eq))
````

### Automatic C-Code Generation

We can generate the C++ implementation **directly from our final difference equation**. There is no need to rewrite the equation or enter the filter coefficients manually.

Let $q=T_s\omega_c$. Solving the equation for the current output gives

$$
\boxed{y_k=\frac{
4(x_k-2x_{k-1}+x_{k-2})
+(8-2q^2)y_{k-1}
-(q^2-2\sqrt{2}q+4)y_{k-2}
}{q^2+2\sqrt{2}q+4}}
$$

This time the input term is $4(x_k-2x_{k-1}+x_{k-2})$, rather than the low-pass expression $q^2(x_k+2x_{k-1}+x_{k-2})$. The denominator is the same as in the second-order low-pass filter.

First, let SymPy solve **our original** `eq` for $y_k$, substitute $q$, and generate its C expression.

````python
q, q2 = symbols('q q2', positive=True)

# Solve the difference equation derived above (not a manually entered formula).
yk = solve(eq, y0)[0]

# Substitute q = Ts*wc, then precompute q**2 in configure().
yk = cancel(yk.subs(Ts, q/wc))
yk = yk.xreplace({q**2: q2})
mprintb(latex(Eq(y0, yk)))

# Rename mathematical subscripts to ordinary C/C++ variable names.
names = {
    x0: Symbol('x0'), x1: Symbol('x1'), x2: Symbol('x2'),
    y1: Symbol('y1'), y2: Symbol('y2')
}

sample = ccode(yk.xreplace(names))
print('const float y0 = ' + sample + ';')
````

We can now insert the generated `sample` expression into an Arduino-compatible class. The class only handles the previous input/output samples; **the filter mathematics comes from SymPy**.

The header is generated automatically as `butterworth_hpf2.h`. As in the low-pass example, `configure(Ts, wc)` expects the sampling period in seconds and the **analog** cutoff angular frequency in rad/s.

````python
from pathlib import Path

header = r'''#pragma once

class ButterworthHPF2 {
    float q = 0.0f, q2 = 0.0f;
    float x1 = 0.0f, x2 = 0.0f;
    float y1 = 0.0f, y2 = 0.0f;

public:
    void configure(float Ts, float wc) {
        q = Ts * wc;
        q2 = q * q;
        reset();
    }

    void reset() {
        x1 = x2 = 0.0f;
        y1 = y2 = 0.0f;
    }

    float step(float x0) {
        const float y0 = __SAMPLE__;
        x2 = x1; x1 = x0;
        y2 = y1; y1 = y0;
        return y0;
    }
};
'''

# SymPy may emit M_SQRT2; use a float literal for portable Arduino C++.
header = header.replace('__SAMPLE__', sample.replace('M_SQRT2', '1.4142135623730951f'))
Path('butterworth_hpf2.h').write_text(header, encoding='utf-8')
print(header)
````

#### Verification with SciPy

As a final check, we compare the coefficients obtained **from our SymPy output equation** against SciPy's standard digital second-order Butterworth HPF. SciPy specifies the **digital** cutoff frequency in Hz, so the analog cutoff supplied to our Tustin equation must first be prewarped:

$$\omega_c=2f_s\tan\!\left(\frac{\pi f_c}{f_s}\right).$$

For example, choose $f_s=1000$ Hz and $f_c=40$ Hz. The five coefficients should agree to floating-point precision.

````python
import numpy as np
from scipy.signal import butter

fs, fc = 1000.0, 40.0
q_value = 2*np.tan(np.pi*fc/fs)
expr = yk.subs({q: q_value, q2: q_value**2})

# A linear difference equation: differentiate to read off the coefficients.
b_sym = np.array([float(diff(expr, xk)) for xk in (x0, x1, x2)])
a_sym = np.array([1.0] + [-float(diff(expr, yk_old)) for yk_old in (y1, y2)])
b_ref, a_ref = butter(2, fc, fs=fs, btype='highpass', output='ba')

print('           SymPy/Tustin      SciPy/butter')
for label, ours, reference in zip(('b0','b1','b2','a0','a1','a2'),
                                  np.r_[b_sym,a_sym], np.r_[b_ref,a_ref]):
    print(f'{label:>2}: {ours: .12f}   {reference: .12f}')

assert np.allclose(b_sym, b_ref, atol=1e-12, rtol=0)
assert np.allclose(a_sym, a_ref, atol=1e-12, rtol=0)
print('PASS: the HPF coefficients agree with SciPy.')
````

### Implementation in Simulink

```octave
function y = HPF(y1, y2, x0, x1, x2, wc, Ts)
% Second-order Butterworth high-pass filter
%
% y1 = y(k-1)
% y2 = y(k-2)
% x0 = x(k)
% x1 = x(k-1)
% x2 = x(k-2)
% wc = cutoff angular frequency [rad/s]
% Ts = sampling period [s]

sq22 = 2*sqrt(2);
Tswc = Ts * wc;

A = 1 + sq22/Tswc + 4/Tswc^2;
B = 2 - 8/Tswc^2;
C = 1 - sq22/Tswc + 4/Tswc^2;
D = 4/Tswc^2;

y = (D*(x0 - 2*x1 + x2) - B*y1 - C*y2) / A;
end
```
